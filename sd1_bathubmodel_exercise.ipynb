```python
## Import packages
import numpy as np
import matplotlib.pyplot as plt
```

```python
## setting the parameters

### flows parameters rate
birth_rate = 0.025
death_rate = 0.01
in_migration_rate = 0.005
out_migration_rate = 0.002

### time parameters
t_start = 0
t_end = 50
dt = 1
```

```python
## setting the initial population

### time array
t = np.arange(t_start, t_end, dt)

### create empty arrrays for population and flows to store the results
population, births, deaths, in_migration, out_migration, flows = np.zeros(len(t)), np.zeros(len(t)), np.zeros(len(t)), np.zeros(len(t)), np.zeros(len(t)), np.zeros(len(t))

### initiate the initial population
population[0] = 1000000
```

```python
## loop simulation

for i in range(1, len(t)):
    # calculate the flows
    births_val = birth_rate * population[i-1]
    deaths_val = death_rate * population[i-1]
    in_migration_val = in_migration_rate * population[i-1]
    out_migration_val = out_migration_rate * population[i-1]
    flows_val = births_val - deaths_val + in_migration_val - out_migration_val
    
    # store the value into the arrays
    population[i] = population [i-1] + flows_val * dt
    births[i] = births_val
    deaths[i] = deaths_val
    in_migration[i] = in_migration_val
    out_migration[i] = out_migration_val
    flows[i] = flows_val
```

```python
## plot the results
plt.plot(t, population)
plt.title('Urban Population Growth (Natural and Migration Components)')
plt.xlabel('Time (years)')
plt.ylabel('Population (millions)')
plt.show()
```

```python
## plotting the flows
plt.plot(t[1:], births[1:], label="Births")
plt.plot(t[1:], deaths[1:], label="Deaths")
plt.plot(t[1:], in_migration[1:], label="In-Migration")
plt.plot(t[1:], out_migration[1:], label="Out-Migration")
plt.plot(t[1:], flows[1:], label="Net Flow", color='black', linewidth=2, linestyle='--')
plt.title('Flows in Urban Population Growth')
plt.xlabel('Time (years)')
plt.ylabel('Flows (population per year)')
plt.legend()
plt.show()
```
