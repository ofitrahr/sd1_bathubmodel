```python
## import package
import numpy as np
import matplotlib.pyplot as plt
```

```python
## setting the parameter
dt = 1.0 # time step
t_end = 50 # simulation time
birth_rate = 0.025 # 2.5% per year
death_rate = 0.015 # 1.5% per year
```

```python
## setting the initial condition
t = np.arange(0, t_end+dt, dt) # time array
population = np.zeros(len(t)) # population array
population[0] = 1000 # initial population
```

```python
## loop for simulation (Euler method)
for i in range(1, len(t)):
    births = birth_rate * population[i-1]
    deaths = death_rate * population[i-1]
    net_flow = np.round(births - deaths)
    population[i] = population[i-1] + net_flow * dt


```

```python
## Plotting the results
plt.plot(t, population, label='Population')
plt.xlabel('Time (years)')
plt.ylabel('Population')
plt.title('Bathub Model: Population Dynamics')
plt.legend()
plt.show()
```
